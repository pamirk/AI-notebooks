## Merge Exercises:

**Exercise**: Read in the child_mortality_rates and adult_mortality_rates datasets and merge them on appropriate variables, using a meaningful merge technique. Perform data cleaning where necessary.

```python
cm = pd.read_csv('../datasets/child_mortality_rates.csv')
cols = dict(zip(cm.columns, ['Country','Year','Data Source','Neonatal F','Neonatal M','Infant F','Infant M','U5 F','U5 M',]))
cm.rename(columns=cols, inplace=True)
cm.drop(0, inplace=True)
cm.head()
```

```python
def remove(x):
    if type(x) == str :
        if x.find(' ') == -1 :
            return x
        else : return x[0: x.find(' ')]    #[ : x.str.find(' ')]
    return x
```

```python
def for_each_col(x):
    return x.apply(remove)
```

```python
cm[cm.columns[3:]] = cm[cm.columns[3:]].apply(for_each_col)    
cm.head()
```

```python
am = pd.read_csv('../datasets/adult_mortality_rates.csv')
am.rename(columns=am.iloc[0], inplace=True)
am.drop(0, inplace=True)
```

```python
am.head()
```

```python
cmam = pd.merge(cm, am, left_on=['Country','Year'], right_on=['Country','Year'])
cmam.head()
```
